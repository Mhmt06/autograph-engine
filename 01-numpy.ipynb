Weil `np.random.seed(0)` gesetzt ist, bleiben die Zufallszahlen immer exakt dieselben. Das ist zwar nicht zwingend notwendig, macht deine Experimente aber verlässlich reproduzierbar. Wenn du deine Daten zum Beispiel zufällig in Trainings- und Testsets aufteilst, würde ohne Seed bei jedem Durchlauf etwas anderes herauskommen – und jemand, der deinen Code ausprobiert, erhielte völlig andere Ergebnisse. Der Seed sorgt dafür, dass die Zahlen zwar zufällig erzeugt werden, dieser Zufall aber jederzeit exakt wiederholbar ist. Genau deshalb spricht man von „Pseudozufall“.

# 4. Manipulating and Comparing Arrays

## Arithmetic
* **Basic Operations:** `+`, `-`, `*`, `/`, `//`, `**`, `%`
* **Mathematical Functions:**
  * `np.exp()` – Exponentialfunktion
  * `np.log()` – Natürlicher Logarithmus
* **Matrix Multiplication:**
  * Dot product (`np.dot()`)
* **Broadcasting:** Automatisches Anpassen von Array-Formen bei Rechenoperationen

> ### 📐 Regeln für Broadcasting
> Wenn zwei Arrays unterschiedliche Formen (Shapes) haben, wendet NumPy folgende 3 Regeln an:
> 1. **Auffüllen mit 1en:** Haben die Arrays unterschiedlich viele Dimensionen, wird die Shape des kleineren Arrays auf der **linken Seite** mit Einsen aufgefüllt (z. B. wird `(3,)` zu `(1, 3)`).
> 2. **Strecken:** Stimmt die Größe in einer Dimension nicht überein, wird die Dimension mit der Größe `1` so weit **gestreckt** (wiederholt), bis sie zur Größe des anderen Arrays passt.
> 3. **Fehler:** Unterscheiden sich die Größen in einer Dimension und **keine** von beiden ist `1`, wirft NumPy einen Fehler (`ValueError`).

---

## Aggregation
> **Tipp:** NumPy-Funktionen (wie `np.sum()`) sind auf NumPy-Arrays wesentlich schneller als die eingebauten Python-Funktionen (wie `sum()`).

* `np.sum()` – Summe aller Elemente
* `np.mean()` – Mittelwert (Durchschnitt)
* `np.std()` – Standardabweichung
* `np.var()` – Varianz
* `np.min()` – Minimum
* `np.max()` – Maximum
* `np.argmin()` – Index des kleinsten Werts finden
* `np.argmax()` – Index des größten Werts finden

> **Achsen-Steuerung (`axis`):**  
> Alle diese Funktionen funktionieren auf allen `ndarray`s. Mit dem `axis`-Parameter kannst du die Richtung angeben:
> ```python
> a4.min(axis=0)  # Berechnet das Minimum entlang der Achse 0
> ```

---

## Reshaping & Transposing
* **Reshaping:** `np.reshape()` – Ändert die Form eines Arrays
* **Transposing:** `a3.T` – Vertauscht Achsen / Zeilen und Spalten

---

## Comparison Operators
Vergleichsoperatoren geben Arrays mit Wahrheitswerten (`True` / `False`) zurück:

* **Operatoren:** `>`, `<`, `<=`, `>=`, `==`, `!=` (z. B. `x != 3`, `x == 3`)
* **Kombination mit Aggregation:**
  * `np.sum(x > 3)` – Zählt, wie viele Elemente größer als 3 sind

# Variance, Standard Deviation & Histograms

## 1. Varianz (`np.var`) & Standardabweichung (`np.std`)
* **Varianz (`np.var`):** Misst, wie weit die Zahlen durchschnittlich von ihrem Mittelwert entfernt liegen (gestreut sind).
* **Standardabweichung (`np.std`):** Ist einfach die **Quadratwurzel der Varianz** (`np.sqrt(np.var())`). Sie gibt die Streuung in der ursprünglichen Einheit der Daten an.

```python
# Die Standardabweichung ist die Wurzel aus der Varianz
np.sqrt(np.var(high_var_array)) == np.std(high_var_array)

# Visualizing Data: Histograms

## What is a Histogram?
A histogram shows the **distribution of a dataset** by grouping numbers into bins and counting how many values fall into each bin.

* **X-axis (Values):** Represents the range of data points (from 0 to 5000).
* **Y-axis (Frequency):** Shows how many numbers belong to that specific range.
